## Import and Installs

```python
import sys
import os
import uuid
```

```python
sys.path.append(os.path.abspath(".."))

from src.config.settings import LakehouseConfig
from src.ingestion.autoloader import ingest_autoloader_stream
```

```python
# Parametrizacao da execucao
config = LakehouseConfig()
current_batch_id = str(uuid.uuid4())

spark.sql(f"USE CATALOG {config.catalog}")
spark.sql(f"USE SCHEMA {config.schema_bronze}")

print(f"Iniciando ingestao bronze. Batch ID: {current_batch_id}")
```

## json

```python
# clientes CDC
ingest_autoloader_stream(
    spark=spark,
    source_format="json",
    source_path=f"{config.raw_path}/clientes_cdc",
    target_table=f"{config.catalog}.{config.schema_bronze}.bronze_clientes_cdc",
    checkpoint_dir=f"{config.checkpoint_base}/clientes_cdc",
    batch_id=current_batch_id
)

# contas CDC
ingest_autoloader_stream(
    spark=spark,
    source_format="json",
    source_path=f"{config.raw_path}/contas_cdc",
    target_table=f"{config.catalog}.{config.schema_bronze}.bronze_contas_cdc",
    checkpoint_dir=f"{config.checkpoint_base}/contas_cdc",
    batch_id=current_batch_id
)

#cartoes CDC
ingest_autoloader_stream(
    spark=spark,
    source_format="json",
    source_path=f"{config.raw_path}/cartoes_cdc",
    target_table=f"{config.catalog}.{config.schema_bronze}.bronze_cartoes_cdc",
    checkpoint_dir=f"{config.checkpoint_base}/cartoes_cdc",
    batch_id=current_batch_id
)
```

## csv


```python
# transacoes
ingest_autoloader_stream(
    spark=spark,
    source_format="csv",
    source_path=f"{config.raw_path}/transacoes",
    target_table=f"{config.catalog}.{config.schema_bronze}.bronze_transacoes",
    checkpoint_dir=f"{config.checkpoint_base}/transacoes",
    batch_id=current_batch_id,
    extra_options={"header": "true", "cloudFiles.partitionColumns": "dt_particao"}
)

# eventos de riscos
ingest_autoloader_stream(
    spark=spark,
    source_format="csv",
    source_path=f"{config.raw_path}/eventos_risco",
    target_table=f"{config.catalog}.{config.schema_bronze}.bronze_eventos_risco",
    checkpoint_dir=f"{config.checkpoint_base}/eventos_risco",
    batch_id=current_batch_id,
    extra_options={"header": "true"}
)

# estornos
ingest_autoloader_stream(
    spark=spark,
    source_format="csv",
    source_path=f"{config.raw_path}/estornos",
    target_table=f"{config.catalog}.{config.schema_bronze}.bronze_estornos",
    checkpoint_dir=f"{config.checkpoint_base}/estornos",
    batch_id=current_batch_id,
    extra_options={"header": "true"}
)

print("Ingestao Bronze concluida com sucesso!")
```
