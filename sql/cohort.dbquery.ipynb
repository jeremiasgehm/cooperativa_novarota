```sql
SELECT
  date_format(data_inicio_vigencia, 'yyyy-MM') AS safra_adesao,
  dim.segmento,
  COUNT(DISTINCT cli.id_cliente) AS cooperados_ativos,
  ROUND(AVG(cli.ticket_medio), 2) AS ticket_medio_safra,
  ROUND(SUM(cli.volume_total_transacionado), 2) AS volume_consolidado_safra
FROM
  novarota.gold.gold_metricas_cooperados cli
    JOIN novarota.silver.dim_cliente_scd2 dim
      ON cli.id_cliente = dim.id_cliente
      AND dim.flag_ativo = TRUE
GROUP BY
  1,
  2
ORDER BY
  safra_adesao DESC,
  volume_consolidado_safra DESC
```
