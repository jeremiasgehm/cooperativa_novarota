```sql
-- distrib de violacao em transações
SELECT 
    _motivo_quarentena,
    COUNT(*) AS total_ocorrencias,
    MIN(_data_quarentena) AS primeira_ocorrencia,
    MAX(_data_quarentena) AS ultima_ocorrencia
FROM novarota.silver.quarentena_transacoes
GROUP BY _motivo_quarentena
ORDER BY total_ocorrencias DESC
```
