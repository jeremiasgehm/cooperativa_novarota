```sql
-- indicadores consolidados por canal de atendimento
SELECT
  canal,
  status_transacao,
  quantidade_eventos,
  ROUND(valor_total_liquido, 2) AS volume_financeiro,
  ROUND(
    (valor_total_liquido / SUM(valor_total_liquido) OVER (PARTITION BY canal)) * 100,
    2
  ) AS percentual_volume_canal
FROM
  novarota.gold.gold_desempenho_canais
ORDER BY
  canal,
  status_transacao
```
