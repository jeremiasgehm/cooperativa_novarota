```sql
SELECT 'bronze_clientes_cdc' AS tabela, COUNT(*) AS total_registos FROM novarota.bronze.bronze_clientes_cdc
UNION ALL
SELECT 'bronze_contas_cdc', COUNT(*) FROM novarota.bronze.bronze_contas_cdc
UNION ALL
SELECT 'bronze_cartoes_cdc', COUNT(*) FROM novarota.bronze.bronze_cartoes_cdc
UNION ALL
SELECT 'bronze_transacoes', COUNT(*) FROM novarota.bronze.bronze_transacoes
```
