```sql
-- segmentos
SELECT 
    segmento,
    COUNT(id_cliente) AS total_cooperados,
    ROUND(SUM(volume_total_transacionado), 2) AS volume_financeiro_total,
    ROUND(AVG(ticket_medio), 2) AS ticket_medio_geral,
    SUM(total_transacoes) AS total_operacoes
FROM novarota.gold.gold_metricas_cooperados
GROUP BY segmento
ORDER BY volume_financeiro_total DESC
```
