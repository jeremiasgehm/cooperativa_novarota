```sql
-- assoc que tiveram alterações cadastrais ao longo do tempo (versões antigas vs. versão atual)
WITH cooperados_com_historico AS (
    SELECT id_cliente
    FROM novarota.silver.dim_cliente_scd2
    GROUP BY id_cliente
    HAVING COUNT(*) > 1
    LIMIT 5
)
SELECT 
    d.id_cliente,
    d.nome,
    d.cidade,
    d.estado,
    d.renda,
    d.segmento,
    d.data_inicio_vigencia,
    d.data_fim_vigencia,
    d.flag_ativo,
    d._hash_diff
FROM novarota.silver.dim_cliente_scd2 d
INNER JOIN cooperados_com_historico h ON d.id_cliente = h.id_cliente
ORDER BY d.id_cliente, d.data_inicio_vigencia ASC
```
